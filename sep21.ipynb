## Fundamental Data Types
### 1.int(integer) :- it is a number without floating point.

```python
a=11
print(a)
print(type(a))
```

### 2.float:-it is a number with decimal position.

```python
price_petrol=111.54
print(price_petrol)
print(type(price_petrol))
      
```

### 3.str:- it is used to store sequence of character or text.

```python
name="Monika"
print(name)
print(type(name))
```

### 4.complex:- for mathematical processing of data we need complex data type.
-c=10+20j
-real part
-imaginary part

```python
a=3*5j
print(a)
print(type(a))
```

### 5.boolean:-is a data type that represents one of two values:True or False.

```python
is_present =True
is_logged_in =False
```
